# Evidence summary

Creates descriptive, non-claiming summaries from the evidence workbook. Empty templates intentionally produce empty summaries.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parents[1]
WORKBOOK = ROOT / "literature" / "evidence_matrix.xlsx"
studies = pd.read_excel(WORKBOOK, sheet_name="Studies", dtype=str).fillna("")
evidence = pd.read_excel(WORKBOOK, sheet_name="Evidence Extraction", dtype=str).fillna("")

def counts(column):
    values = studies.loc[studies[column].ne(""), column]
    return values.value_counts().rename_axis(column).to_frame("study_count")

print(f"Studies recorded: {len(studies):,}")
print(f"Evidence findings recorded: {len(evidence):,}")
for field in ["Topic", "Estimation Method", "Fault Type", "Platform Type", "Validation Level"]:
    print(f"\n{field}")
    display(counts(field))
```
